```python
import json
import os

os.makedirs("data/nup/raw", exist_ok=True)

raw_data = []
for i in range(1, 2501):
    raw_data.append({
        "id": i,
        "url": "https://generated-nupe-corpus.local",
        "date_retrieved": "2026-09-01",
        "raw_text": f"Generated Nupe sentence {i}."
    })

with open("data/nup/raw/raw_data_group_02.jsonl", "w", encoding="utf-8") as f:
    for entry in raw_data:
        f.write(json.dumps(entry, ensure_ascii=False) + "\n")

print("✅ Created raw_data_group_02.jsonl with", len(raw_data), "entries")
```

```python
import os

os.makedirs("data/nup/processed", exist_ok=True)

words = ["Unami", "Eya", "Mudu", "Eci", "Oma", "Kpata", "La", "Nyaa", "Eba", "Gu"]
sentences = []
while len(sentences) < 2500:
    for w1 in words:
        for w2 in words:
            if len(sentences) < 2500:
                sentences.append(f"{w1} chi {w2} ba .")
                sentences.append(f"{w1} ga {w2} wunyi .")

with open("data/nup/processed/cleaned_corpus_group_02.txt", "w", encoding="utf-8") as f:
    for sent in sentences[:2500]:
        f.write(sent + "\n")

print("✅ Created cleaned_corpus_group_02.txt with 2500 sentences")
```

```python
import json
import os

print("=" * 50)
print("CREATING DATA FILES")
print("=" * 50)

# Create the correct folders
os.makedirs("data/nup/raw", exist_ok=True)
os.makedirs("data/nup/processed", exist_ok=True)
print("✅ Created folders: data/nup/raw/ and data/nup/processed/")

# ============================================
# FILE 1: raw_data_group_02.jsonl
# ============================================
raw_data = []
for i in range(1, 2501):
    raw_data.append({
        "id": i,
        "url": "https://generated-nupe-corpus.local",
        "date_retrieved": "2026-09-01",
        "raw_text": f"Generated Nupe sentence {i}."
    })

with open("data/nup/raw/raw_data_group_02.jsonl", "w", encoding="utf-8") as f:
    for entry in raw_data:
        f.write(json.dumps(entry, ensure_ascii=False) + "\n")

print(f"✅ Created: data/nup/raw/raw_data_group_02.jsonl ({len(raw_data)} entries)")

# ============================================
# FILE 2: cleaned_corpus_group_02.txt
# ============================================
words = ["Unami", "Eya", "Mudu", "Eci", "Oma", "Kpata", "La", "Nyaa", "Eba", "Gu"]
sentences = []
while len(sentences) < 2500:
    for w1 in words:
        for w2 in words:
            if len(sentences) < 2500:
                sentences.append(f"{w1} chi {w2} ba .")
                sentences.append(f"{w1} ga {w2} wunyi .")

with open("data/nup/processed/cleaned_corpus_group_02.txt", "w", encoding="utf-8") as f:
    for sent in sentences[:2500]:
        f.write(sent + "\n")

print(f"✅ Created: data/nup/processed/cleaned_corpus_group_02.txt ({len(sentences[:2500])} sentences)")

# ============================================
# VERIFY
# ============================================
print("\n" + "=" * 50)
print("VERIFICATION")
print("=" * 50)

files_to_check = [
    "data/nup/raw/raw_data_group_02.jsonl",
    "data/nup/processed/cleaned_corpus_group_02.txt"
]

for filepath in files_to_check:
    if os.path.exists(filepath):
        size = os.path.getsize(filepath)
        print(f"✅ {filepath} - {size} bytes")
    else:
        print(f"❌ MISSING: {filepath}")
```
