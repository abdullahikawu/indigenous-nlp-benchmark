### Laplace Smoothing Formula
$$P(w_2|w_1) = \frac{count(w_1, w_2) + 1}{count(w_1) + V}$$

Where:
- `count(w1, w2)` = frequency of bigram (w1, w2)
- `count(w1)` = frequency of unigram w1
- `V` = vocabulary size

### Perplexity Formula
$$PP = \exp\left( -\frac{1}{N} \sum_{i=1}^{N} \log P(w_i | w_{i-1}) \right)$$
